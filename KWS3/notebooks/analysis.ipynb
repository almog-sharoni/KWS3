# Analysis of Ablation Study Results

This notebook is used to analyze the results obtained from the ablation study conducted on the Keyword Spotting model. It includes visualizations and statistical analysis of the performance metrics.

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set(style='whitegrid')

# Load the ablation results
results_df = pd.read_csv('../ablation_results/ablation_results.csv')

# Display the first few rows of the dataframe
results_df.head()
```

```python
# Plotting test accuracy against different configurations
plt.figure(figsize=(12, 6))
sns.barplot(data=results_df, x='parameter', y='test_accuracy', hue='value')
plt.title('Test Accuracy by Parameter Configuration')
plt.xlabel('Parameter')
plt.ylabel('Test Accuracy (%)')
plt.xticks(rotation=45)
plt.legend(title='Value')
plt.tight_layout()
plt.show()
```

```python
# Save the analysis results
analysis_results = results_df.groupby('parameter').agg({'test_accuracy': 'mean'}).reset_index()
analysis_results.to_csv('analysis_results.csv', index=False)
print('Analysis results saved to analysis_results.csv')
```
